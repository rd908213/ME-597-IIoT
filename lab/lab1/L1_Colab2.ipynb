### Data plotting

Plot a chart that includes temperature (in &deg;C, &deg;F) in the code box below:

```python
df['time'] = pd.to_datetime(df["Timestamp"])
time = df['time'].dt.hour*3600 + df['time'].dt.minute*60 + df['time'].dt.second
t1 = time - time[0]
temperatureC = df['Temperature (°C)']

#Code section of the random number generator with plotting
fig = plt.figure() # Creates a new figure

ax = fig.add_subplot(111) # add a subplot to the new figure, 111 means "1x1 grid, first subplot"
#fig.subplots_adjust(top=0.80) # adjust the placing of subplot, adjust top, bottom, left and right spacing
ax.set_title('Temperature graph',fontsize= 20) # title of plot

ax.set_xlabel('seconds',fontsize = 10) #xlabel
ax.set_ylabel('Degree of Celsius', fontsize = 10) #ylabel


ax.plot(t1,temperatureC,'-o') #plotting the data with marker '-o'
#ax.axis([0, 10, 0, 10]) #specifying plot axes lengths
plt.show()
                      # prints the graph
```

### Task 2.1

Plot a chart using that includes both temperature unit in the code box below:

```python
#Insert your coding below this line
df['time'] = pd.to_datetime(df["Timestamp"])
time = df['time'].dt.hour*3600 + df['time'].dt.minute*60 + df['time'].dt.second
t1 = time - time[0]
temperatureC = df['Temperature (°C)']
temperatureF = df['Temperature (°F)']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5)) # Creates a new figure with 2 subplots side by side
fig.suptitle('Temperature graphs in Celsius and Fahrenheit', fontsize=20) # overall title for the figure

ax1.set_title('Temperature graph (°C)',fontsize= 12) # title of plot
ax1.set_xlabel('seconds',fontsize = 10) #xlabel
ax1.set_ylabel('Degree of Celsius', fontsize = 10) #ylabel

ax2.set_title('Temperature graph (°F)',fontsize= 12) # title of plot
ax2.set_xlabel('seconds',fontsize = 10) #xlabel
ax2.set_ylabel('Degree of Fahrenheit', fontsize = 10) #ylabel

ax1.plot(t1,temperatureC,'-o') #plotting the data with marker '-o'
ax2.plot(t1,temperatureF,'-o') #plotting the data with marker '-o'
#ax.axis([0, 10, 0, 10]) #specifying plot axes lengths
plt.show()
                      # prints the graph
```
