```
import datetime
import time
import random
import pymysql.cursors

## Credential
HOST = 'mepotrb16.ecn.purdue.edu' # MySQL server host DNS
PORT = 3306 # MySQL server port number
USER = 'yourname' # MySQL account name
PASSWORD = 'password' # Password of the account
DB = 'ME597Spring24' # DB name (you should double check your database name)
TABLE = 'yourname_lab6' # table name
## Credential

sensor = 'sensor3' # sensor name
measurement1 = 'acceleration1' # measurement name 1
measurement2 = 'acceleration2' # measurement name 2
measurement3 = 'acceleration3' # measurement name 3

connection = pymysql.connect(host=HOST, user=USER, password=PASSWORD, db=DB, port=PORT) # make a connection to MySQL server
cursor = connection.cursor() # Open cursur to execute SQL query

duration = 60 # collection time in second unit
start_time = time.time() # get current time

while time.time() - start_time < duration:
    value1 = str(random.uniform(0,2)) # random value for measurement1
    value2 = str(random.uniform(-2,0)) # random value for measurement2
    value3 = str(random.uniform(9.7,9.9)) # random value for measurement3
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S') # get datetime as MySQL timestamp format
    
    print(timestamp)
    print("{}={}".format(measurement1,value1)) # value1
    print("{}={}".format(measurement2,value2)) # value2
    print("{}={}".format(measurement3,value3)) # value2
    
    query1 = "INSERT INTO "+TABLE+" (timestamp,sensor,measurement,value) VALUE('"+timestamp+"','"+sensor+"','"+measurement1+"','"+value1+"');" #SQL query1 for value1
    query2 = "INSERT INTO "+TABLE+" (timestamp,sensor,measurement,value) VALUE('"+timestamp+"','"+sensor+"','"+measurement2+"','"+value2+"');" #SQL query2 for value2
    query3 = "INSERT INTO "+TABLE+" (timestamp,sensor,measurement,value) VALUE('"+timestamp+"','"+sensor+"','"+measurement3+"','"+value3+"');" #SQL query3 for value3
    
    cursor.execute(query1) # execute query 1
    cursor.execute(query2) # execute query 2
    cursor.execute(query3) # execute query 3
    connection.commit() # commit all SQL queries
    
    print("==INSERT QUERIES DONE==\n")
    
    time.sleep(3) # wait for 3 seconds

connection.close() # gently close connection
print("==Program DONE==")

```

````python
# Paste your code below.
#

import datetime
import time
import random
import mysql.connector # Using mysql.connector since it is the official MySQL driver for Python and is widely used in the industry. It provides a simple and efficient way to connect to MySQL databases, execute queries, and manage database connections. Additionally, mysql.connector is compatible with Python 3 and offers support for various MySQL features, making it a reliable choice for database interactions in Python applications.

# From lab5_DS18B20_sample.py
import os
import glob

# Taken from lab5_ADXL345_sample.py
import board
import busio
import adafruit_adxl34x
import datetime

class DS18B20:
    def __init__(self):        
        self.sensor = "DS18B20"
        os.system('modprobe w1-gpio')
        os.system('modprobe w1-therm')
        self.base_dir = '/sys/bus/w1/devices/'
        self.device_folder = glob.glob(self.base_dir + '28*')[0]
        self.device_file = self.device_folder + '/w1_slave'

    def read_temp_raw(self):
        f = open(self.device_file, 'r')
        lines = f.readlines()
        f.close()
        return lines

    def read_temp(self):
        lines = self.read_temp_raw()
        while lines[0].strip()[-3:] != 'YES':
            time.sleep(0.2)
            lines = self.read_temp_raw()
        equals_pos = lines[1].find('t=')
        if equals_pos != -1:
            temp_string = lines[1][equals_pos+2:]
            temp_c = float(temp_string) / 1000.0
            return temp_c
    
    def to_db(self, cursor, timestamp):
        temp_c = self.read_temp()
        params = {
            'table': TABLE,
            'timestamp': timestamp,
            'sensor': self.sensor,
            'measurement': 'Temp',
            'value': str(temp_c) + " °C"
        }
        cursor.execute(QUERY, params)
        print("{}={}".format(params['measurement'],params['value']))

class ADXL345:
    def __init__(self):
        self.sensor = "ADXL345"
        self.i2c = busio.I2C(board.SCL, board.SDA)
        self.acc = adafruit_adxl34x.ADXL345(self.i2c)
    
    def read_acceleration(self):
        x_acc, y_acc, z_acc = self.acc.acceleration
        return x_acc, y_acc, z_acc
    
    def to_db(self, cursor, timestamp):
        x_acc, y_acc, z_acc = self.read_acceleration()
        params = {
            'table': TABLE,
            'timestamp': timestamp,
            'sensor': self.sensor,
        }
        
        # X Acceleration
        params['measurement'] = 'Xacc'
        params['value'] = str(x_acc) + " m/s2"
        cursor.execute(QUERY, params)
        print("{}={}".format(params['measurement'],params['value']))
        
        # Y Acceleration
        params['measurement'] = 'Yacc'
        params['value'] = str(y_acc) + " m/s2"
        cursor.execute(QUERY, params)
        print("{}={}".format(params['measurement'],params['value']))
        
        # Z Acceleration
        params['measurement'] = 'Zacc'
        params['value'] = str(z_acc) + " m/s2"
        cursor.execute(QUERY, params)
        print("{}={}".format(params['measurement'],params['value']))

class VirtualSensor:
    def __init__(self):
        self.sensor = "VirtualSensor"
    
    def to_db(self, cursor, timestamp):
        value = random.uniform(0, 100)
        params = {
            'table': TABLE,
            'timestamp': timestamp,
            'sensor': self.sensor,
            'measurement': 'Humd',
            'value': str(value) + " RH%"
        }
        cursor.execute(QUERY, params)
        print("{}={}".format(params['measurement'],params['value']))

## Credential
HOST = '10.165.67.146' # MySQL server host DNS
PORT = 3306 # MySQL server port number
USER = 'robertclaud' # MySQL account name
PASSWORD = 'Qwer1234!' # Password of the account
DB = 'ME597Spring26O' # DB name
TABLE = 'robertclaud_lab6' # table name
QUERY = f"""
    INSERT INTO {TABLE} (
        timestamp,
        sensor,
        measurement,
        value
    ) VALUES (
        %(timestamp)s,
        %(sensor)s,
        %(measurement)s,
        %(value)s
    );
"""

connection = mysql.connector.connect(
    host=HOST,
    user=USER,
    password=PASSWORD,
    db=DB,
    port=PORT
) # make a connection to MySQL server

cursor = connection.cursor(dictionary=True) # Open cursur to execute SQL query

# Set to 120 per (a)
duration = 120 # collection time in second unit
start_time = time.time() # get current time

while time.time() - start_time < duration:
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S') # get datetime as MySQL timestamp format
    
    acc_sensor = ADXL345()
    acc_sensor.to_db(cursor, timestamp)

    temp_sensor = DS18B20()
    temp_sensor.to_db(cursor, timestamp)

    virtual_sensor = VirtualSensor()
    virtual_sensor.to_db(cursor, timestamp)
    connection.commit()
    print("==INSERT QUERIES DONE==\n")
    
    time.sleep(3) # wait for 3 seconds per (b)

connection.close() # gently close connection
print("==Program DONE==")

#
````
