## 3.1 Project solution

[Show and explain your solution. Interpret the output and show how the proposed solution addressed the problem/project. Identify limitations along with ranges of operation.]


---

<img src="https://raw.githubusercontent.com/rd908213/ME-597-IIoT/refs/heads/main/group_project/img/Final%20System%20Diagram.png" width=100%>

The final solution consists of an ADXL345 Accelerometer and a DS18B20 Thermocouple attached to the coffee maker in strategic places: the ADXL345 firmly secured to the side of the machine and the DS18B20 secured to the side of the pot. These two sensors send raw information to the raspberry pi. The raspberry pi takes the raw acceleration data and sends it to a machine learning model trained to determine one of three states:
  1. OFF, UNAVAILABLE - The coffee maker is idle, not brewing any coffee
  2. ON, BREWING - The coffee maker is in normal operation but won't be done for some time
  3. ON, FINISHING - The coffee maker is finishing up and the coffee will be ready to drink soon

The temperature sensor is included to judge how fresh or warm the coffee is before the user drinks it.

The raw acceleration and temperature data as well as the state predictions of the model are sent through MTConnect to a MySQL database. This can then be viewed on a Grafana dashboard which can give the user insight into the current state of the coffee pot and how fresh the coffee is.

The limitations of this project are as follows.
  1. The thermocouple is connected directly to the coffee pot. This can make the coffee difficult to pour and means that the pot must be in close proximity to the machine
  2. The electronics are exposed. This means that if a spill were to occur, the electronics might get damaged.
  3. The ML model is biased towards the ON, BREWING condition rather than the ON, FINISHING condition. This means that the incorrect state can be displayed at times.

Overall, the proposed solution does in fact give insight into the status of a coffee maker with good accuracy and allows the user to move away and do other tasks without needing to watch the pot. This allows the coffee to be drunk at optimal freshness and improves the user experience.
